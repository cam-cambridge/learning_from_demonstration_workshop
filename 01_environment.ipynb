# 1. Set up the environment

Use this notebook on each machine that will collect data, train, or deploy. The commands below target macOS/Linux and use `uv` to create a Python 3.12 virtual environment. Install Git first. Internet access and disk space are needed for packages and model weights.

These commands are for an **interactive terminal** opened at this repository's root. Activating an environment in a notebook cell does not change the notebook kernel or later terminal sessions.

## Install uv

In a terminal on macOS or Linux, install uv using the [official standalone installer](https://docs.astral.sh/uv/getting-started/installation/):

```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

This downloads and runs a script from Astral; review the installer or use a package manager if preferred. For example, with Homebrew, run `brew install uv` instead. Open a new terminal if necessary, then verify the installation:

```bash
uv --version
```

If `uv` is not found, follow the installer's instructions to add its binary directory to your `PATH` and restart the terminal.
